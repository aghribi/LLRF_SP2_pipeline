```python
# Setup - Libraries and Configuration

import pickle
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Configure visualization
plt.style.use('default')
sns.set_palette("husl")
%matplotlib inline

print("✓ Libraries loaded")
```

```python
# Load Processed Data

COOKED_DATA = Path('/sps/m4cast/_spiral2_data/_llrf_data/cooked_data')
STEP_DIR = COOKED_DATA / 'step_XX_name'  # CHANGE THIS

# Load pickle file
pkl_file = STEP_DIR / 'output.pkl'  # CHANGE THIS

if pkl_file.exists():
    with open(pkl_file, 'rb') as f:
        data = pickle.load(f)
    print(f"✓ Loaded {pkl_file.name}")
    print(f"  Keys: {list(data.keys()) if isinstance(data, dict) else 'Not a dict'}")
else:
    print(f"⚠️  File not found: {pkl_file}")
    print("   Run the pipeline first: ./run_pipeline.sh")
```
